```python
import os
import shutil

def extract_pngs(source_folder, destination_folder):
    """
    Recursively walks through 'source_folder' and copies all PNG files found
    into 'destination_folder'.
    """
    os.makedirs(destination_folder, exist_ok=True)

    for root, dirs, files in os.walk(source_folder):
        for file in files:
            if file.lower().endswith('.png'):
                src_path = os.path.join(root, file)
                dst_path = os.path.join(destination_folder, file)
                
                shutil.copy2(src_path, dst_path)
                print(f"Copied {src_path} to {dst_path}")
```

```python
extract_pngs('./data/TCGA@Focus/artifact_free/', './data/training/artifact_free')
```
